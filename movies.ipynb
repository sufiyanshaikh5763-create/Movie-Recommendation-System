## Preprocessing

```python
import ast

def genres_extract(cols):
    L=[]
    for i in ast.literal_eval(cols):
        L.append(i['name'])
    return L
```

```python
data['genres']=data['genres'].apply(genres_extract)
```

```python
data.sample(3)
```

```python
def keyword_extract(cols):
    K=[]
    for i in ast.literal_eval(cols):
        K.append(i['name'])
    return K
```

```python
data['keywords']=data['keywords'].apply(keyword_extract)
```

```python
data.tail()
```

```python
## Extracting only the release year from the release date

data['release_date']=pd.to_datetime(data['release_date'])
```

```python
data['release_date']=data['release_date'].dt.year
```

```python
data.head()
```

```python
## processing the cast column too
def cast(cols):
    L=[]
    j=0
    for i in ast.literal_eval(cols):
        if j<3:
            L.append(i['name'])
            j+=1
    return L
```

```python
data['cast']=data['cast'].apply(cast)
```

```python
data.head()
```

```python
## Crew column preprocessing
data['crew'][0]
```

```python
def crew(cols):
    L=[]
    for i in ast.literal_eval(cols):
        if i['job'] == "Director":
            L.append(i['name'])

    return L
```

```python
data['crew']=data['crew'].apply(crew)
```

```python
data['crew'][1]
```

```python
data['overview']=data['overview'].apply(lambda x:x.split())
```

```python
data.head()
```

```python
data['cast']=data['cast'].apply(lambda x:[i.replace(" ","") for i in x])
data['crew']=data['crew'].apply(lambda x:[i.replace(" ","") for i in x])
data['genres']=data['genres'].apply(lambda x:[i.replace(" ","") for i in x])
data['keywords']=data['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
data.head()
```

```python
data['tags']=data['genres']+data['keywords']+data['overview']+data['cast']+data['crew']
```

```python
data.head()
```

```python
data=data.drop(columns=['genres','keywords','overview','release_date','cast','crew'])
```

## Stemming

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps=PorterStemmer()
```

```python
def stemming(text):
    L=[]
    for i in text.split():
        L.append(ps.stem(i))
    return " ".join(L)
```

```python
data['tags']=data['tags'].apply(stemming)
```

```python
data.head()
```

```python
data['tags'][0]
```

## Vectorization

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features=5000,stop_words='english')

vectors=cv.fit_transform(data['tags'])
vectors=vectors.toarray()
```

```python
vectors
```

```python
cv.get_feature_names_out()[:200]
```

## Cosine Similarity

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity=cosine_similarity(vectors)
print(similarity)
```

```python
data[data['title']=='avatar'].index[0]
```

## Defining the recommend function

```python
def recommend(movie):
    movie_index=data[data['title']==movie].index[0]
    recommendations=similarity[movie_index]
    movie_list=sorted(enumerate(recommendations),reverse=True,key=lambda x:x[1])[1:6]
    for i in movie_list:
        print(data.iloc[i[0]].title)
    
```
